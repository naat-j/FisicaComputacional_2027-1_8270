## 1. La constante de Madelung

En física de la materia condensada, la constante de Madelung da el potencial eléctrico total que siente un átomo en un sólido; y depende de las cargas de los otros átomos cercanos y de sus ubicaciones.

Por ejemplo, el cristal de cloruro de sodio sólido (la sal de mesa), tiene átomos dispuestos en una red cúbica, con átomos de sodio y cloro alternados, teniendo los de sodio una carga positiva \(+e\) y los de cloro una negativa \(-e\), donde \(e\) es la carga del electrón.

Si etiquetamos cada posición en la red con tres coordenadas enteras \((i,j,k)\), entonces los átomos de sodio caen en posiciones donde \(i+j+k\) es par, y los átomos de cloro en posiciones donde \(i+j+k\) es impar.

Consideremos un átomo de sodio en el origen, es decir, \(i=j=k=0\), y calculemos la constante de Madelung....


Escribe un programa para calcular e imprimir la constante de Madelung para el cloruro de sodio. Utiliza un valor de \(L\) tan grande como puedas, sin dejar que tu programa se ejecute en un tiempo razonable (un minuto o menos).

```python
import math

L = 100

M = 0

for i in range(-L, L + 1):
    for j in range(-L, L + 1):
        for k in range(-L, L + 1):

            if i == 0 and j == 0 and k == 0:
                continue

            distancia = math.sqrt(i**2 + j**2 + k**2)

            if (i + j + k) % 2 == 0:
                signo = 1
            else:
                signo = -1

            M = M + signo / distancia

print("La constante de Madelung es:", M)
```

## 3. Coeficientes binomiales



### a)

Utiliza esta fórmula para escribir una función llamada `binomial(n,k)` (o como tú quieras) que calcule el coeficiente binomial para un \(n\) y \(k\) dados.

Asegúrate de que tu función devuelva la respuesta en forma de un número entero (no flotante) y proporcione el valor correcto de 1 para el caso en que \(k=0\).

```python
def binomial(n, k):
    
    if k == 0:
        return 1
    
    resultado = 1
    
    for i in range(1, k + 1):
        resultado = resultado * (n - i + 1) // i
    
    return resultado


print(binomial(8, 0))
```

Escribe un programa para calcular:

1. La probabilidad total de que una moneda lanzada 100 veces salga águila exactamente 60 veces.
2. La probabilidad de que salga águila 60 veces o más.

```python
def binomial(n, k):
    
    if k == 0:
        return 1
    
    resultado = 1
    
    for i in range(1, k + 1):
        resultado = resultado * (n - i + 1) // i
    
    return resultado


n = 100

# Probabilidad de obtener exactamente 60 águilas
k = 60

probabilidad_60 = binomial(n, k) / 2**n

print("Probabilidad de obtener exactamente 60 águilas:", probabilidad_60)


# Probabilidad de obtener 60 águilas o más
probabilidad_60_o_mas = 0

for k in range(60, 101):
    probabilidad_60_o_mas = probabilidad_60_o_mas + binomial(n, k) / 2**n

print("Probabilidad de obtener 60 águilas o más:", probabilidad_60_o_mas)
```

Es posible escribir un programa bastante rápido para calcular números primos siguiendo las siguientes observaciones:

**a)** Un número \(n\) es primo si no tiene factores primos menores que \(n\). Por lo tanto, solo necesitamos comprobar si es divisible por otros números primos.

Escribe un programa que encuentre todos los números primos hasta 10 000.

...

Puedes imprimir la lista completa al final del programa, o bien, ir imprimiendo los números individualmente a medida que los vayas encontrando.

```python
import math

primos = [2]

for n in range(3, 10001):

    es_primo = True

    for primo in primos:

        if primo > math.sqrt(n):
            break

        if n % primo == 0:
            es_primo = False
            break

    if es_primo:
        primos.append(n)

print(primos)
```

### a) Números de Catalan

Anteriormente, en clase encontramos los números de Catalán \(C_n\). Si hacemos una pequeña modificación, la definición de \(C_n\) se puede reescribir de la siguiente forma:

\[
C_n=
\begin{cases}
1, & \text{si } n=0,\\[4pt]
\frac{4n-2}{n+1}C_{n-1}, & \text{si } n>0.
\end{cases}
\]

Escribe una función, usando recursividad, que calcule \(C_n\) y utilízala para calcular e imprimir \(C_{100}\).

```python
def catalan(n):

    if n == 0:
        return 1

    else:
        return (4*n - 2) * catalan(n - 1) // (n + 1)


C100 = catalan(100)

print("C100 =", C100)
```

### b) Algoritmo de Euclides

Euclides demostró que el máximo común divisor \(g(m,n)\) de dos enteros no negativos \(m\) y \(n\) satisface que

\[
g(m,n)=
\begin{cases}
m, & \text{si } n=0,\\[4pt]
g(n,m\bmod n), & \text{si } n>0.
\end{cases}
\]

Usando esta fórmula, escribe una función `g(m,n)` que emplee recursividad para calcular el máximo común divisor de \(m\) y \(n\).

Usa tu función para calcular e imprimir el máximo común divisor de 108 y 192.

```python
def g(m, n):

    if n == 0:
        return m

    else:
        return g(n, m % n)


resultado = g(108, 192)

print("El máximo común divisor es:", resultado)
```
