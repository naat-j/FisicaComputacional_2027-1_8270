# Regla de Simpson

También vimos en clase que podríamnos mejorar nuestra aproximación si en vez de usar una linea recta para ajustar la función entre los extremos de cada partición, usamos una curva P(x), por ejemplo una parábola. Y derivamos la regla de Simpson como una mejor forma de aproximar la integral:

```python


def simpson(f, a, b, N):
        h = (b-a)/N
        suma = f(a) + f(b)
        for k in range(1, N//2 + 1):
            x = a + (2*k-1)*h
            suma = suma + 4*f(x)

        for k in range (1, N//2):
            x = a + 2*k*h
            suma = suma + 2*f(x)

        integral = (h/3)*suma

        return integral

    
```
